```python
import gensim.downloader as api
import pandas as pd

GAMES_DATA = "../data/clean_data.csv"
w2v_model = api.load('word2vec-google-news-300')
games_data = pd.read_csv(GAMES_DATA, sep=";")
games_data
```

```python
games_data['tags_token'] = games_data['Tags'].apply(lambda x: x.split(','))
games_data.tags_token
```

```python
import numpy as np
def get_mean_embedding(tokens, model):
    embeddings = [model[word] for word in tokens if word in model]
    if embeddings:
        return np.mean(embeddings, axis=0)
    else:
        return np.zeros(model.vector_size)

games_data['tags_embd'] = games_data['tags_token'].apply(lambda x: get_mean_embedding(x, w2v_model))
```
